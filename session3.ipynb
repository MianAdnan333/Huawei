```python
year = int(input("Enter a year: "))
if year%4 == 0:
    if year%100 == 0:
        if year%400 == 0:
            print(f"{year} is a leap year")
        else:
            print(f"{year} is not a leap year")
    else:
        print(f"{year} is a leap year")
else:
    print(f"{year} is not a leap year")
```

```python
try:
    name = input("Patient name: ")

    age = int(input("Age: "))

    pain = int(input("Pain (0-10): "))

    temperature = float(input("Temperature (C): "))

    score = 0

    if pain >= 8:
        score = score + 10

    elif pain >= 5:
        score = score + 5

    if temperature > 39:
        score = score + 5

    if age > 65 or age < 5:
        score = score + 2

    if score >= 12:
        category = "CRITICAL"

    elif score >= 6:
        category = "URGENT"

    else:
        category = "STANDARD"

    print(name, "scored", score, category)


except ValueError:
    print("Invalid input! Please enter numbers where required.")
```

```python
home_city = "Lahore"

trusted_merchants = {"metro", "ptcl", "shell"}

transactions = [
    ("metro", 42, "Lahore"),
    ("shell", 30, "Lahore"),
    ("luxwatch", 5000, "Dubai"),
    ("cafe99", 18, "Karachi")
]

for transaction in transactions:

    merchant, amount, city = transaction

    flags = 0

    if amount > 1000:
        flags = flags + 1

    if city != home_city:
        flags = flags + 1

    if merchant not in trusted_merchants:
        flags = flags + 1

    if flags >= 3:
        verdict = "BLOCKED"
    elif flags >= 1:
        verdict = "REVIEW"
    else:
        verdict = "OK"

    print(merchant, "$", amount, city, "->", flags, verdict)
```
